Press 4 weekly output 

Daily productuion of stamping press 4, monday to friday 

```python
units = [428, 445, 430, 470, 490]
total = sum(units)
print("Weekly total:", total)
```

```python
import matplotlib.pyplot as plt 

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
plt.bar(days, units)
plt.title('press 4 output')
plt.ylabel('Units')
plt.show()
```
